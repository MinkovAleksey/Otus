# III. Visualization

### Постройте гистограмму распределения признака 'total_bill'

```python
plt.style.use('seaborn-v0_8')
fig, axes = plt.subplots(1, 2, figsize=(15, 6))

# Гистограмма с seaborn
sns.histplot(data=tips, x='total_bill', bins=20, ax=axes[0], kde=True)
axes[0].set_title('Распределение total_bill (Seaborn)')
axes[0].set_xlabel('Total Bill')
axes[0].set_ylabel('Частота')


# Гистограмма с matplotlib
axes[1].hist(tips['total_bill'], bins=20, color='skyblue', edgecolor='black', alpha=0.7)
axes[1].axvline(tips['total_bill'].mean(), color='red', linestyle='--', linewidth=2, label=f'Среднее: {tips["total_bill"].mean():.2f}')
axes[1].axvline(tips['total_bill'].median(), color='green', linestyle='--', linewidth=2, label=f'Медиана: {tips["total_bill"].median():.2f}')
axes[1].set_title('Распределение total_bill (Matplotlib)')
axes[1].set_xlabel('Total Bill')
axes[1].set_ylabel('Частота')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

### Постройте scatterplot, представляющий взаимосвязь между признаками 'total_bill' и 'tip'

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=tips,
    x='total_bill',
    y='tip',
    hue='time',
    style='sex', 
    s=100,
    alpha=0.7 
)

plt.title('Взаимосвязь между total_bill и tip', fontsize=16)
plt.xlabel('Сумма счёта (total_bill)', fontsize=12)
plt.ylabel('Чаевые (tip)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)

plt.show()
```

### Постройте два графика scatterplot, представляющих взаимосвязь между признаками 'total_bill' и 'tip' один для Male, другой для Female и раскрасьте точки в зависимоти от признака 'smoker'

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
sns.scatterplot(
    data=tips[tips['sex'] == 'Male'],
    x='total_bill',
    y='tip',
    hue='smoker',
    palette={True: 'red', False: 'blue'}
)
plt.title('Male: Total Bill vs Tip')


plt.subplot(1, 2, 2)
sns.scatterplot(
    data=tips[tips['sex'] == 'Female'],
    x='total_bill',
    y='tip',
    hue='smoker',
    palette={True: 'red', False: 'blue'}
)
plt.title('Female: Total Bill vs Tip')

plt.tight_layout()
plt.show()
```
